```python
import duckdb
import matplotlib.pyplot as plt
```

```python
con = duckdb.connect("C:/Users/stijn/OneDrive/Documenten/GitHub/DataEngineering_M_S/dvdrental/dvdrental.duckdb", read_only=True)
```

```python
# top 15 steden voor de leesbaarheid (anders worden het er ~~600)
top15 = df.head(15)

plt.figure(figsize=(10, 6))
plt.barh(top15["stad"], top15["gemiddelde_opbrengst_per_klant"])
plt.gca().invert_yaxis()  # zodat de hoogste bovenaan staat
plt.xlabel("Gemiddelde opbrengst per klant (€)")
plt.title("Top 15 steden — gemiddelde opbrengst per klant")
plt.tight_layout()
plt.show()
```
