```python
import pandas
import matplotlib.pyplot as plt
import duckdb
```

```python
con= duckdb.connect('C:/Users/stijn/OneDrive/Documenten/GitHub/DataEngineering_M_S/dvdrental/dvdrental.duckdb')
```

```python

# query
query= """ SELECT
    SUM(p.amount) AS totale_opbrengst
FROM payment p
JOIN customer c ON p.customer_id = c.customer_id
JOIN address ca ON c.address_id = ca.address_id
JOIN city cc ON ca.city_id = cc.city_id
JOIN store s ON TRUE
JOIN address sa ON s.address_id = sa.address_id
JOIN city sc ON sa.city_id = sc.city_id
WHERE cc.city_id = sc.city_id;
"""


df = con.execute(query).fetchdf()

print(df)
```
