```python

# query
query= """ SELECT
    SUM(p.amount) AS totale_opbrengst
FROM payment p
JOIN customer c ON p.customer_id = c.customer_id
JOIN address ca ON c.address_id = ca.address_id
JOIN city cc ON ca.city_id = cc.city_id
JOIN store s ON TRUE
JOIN address sa ON s.address_id = sa.address_id
JOIN city sc ON sa.city_id = sc.city_id
WHERE cc.city_id = sc.city_id;
"""


df = con.execute(query).fetchdf()

print(df)
```

### Binnen de dvd rental data set zijn er geen enkele klanten die een opbrengst leveren die uit dezelfde stad komen als de stad waar de winkel staat

```python
con = duckdb.connect("C:/Users/stijn/OneDrive/Documenten/GitHub/DataEngineering_M_S/dvdrental/dvdrental.duckdb", read_only=True)
```

```python
df = con.execute("SELECT * FROM Fact_opbrengst_winkelstad_vergelijking").df()
```

```python
con.close()
```

```python
groepen = ["Zelfde stad als winkel", "Andere stad"]
df = df.set_index("groep").reindex(groepen, fill_value=0).reset_index()

plt.figure(figsize=(7, 5))

bars = plt.bar(df["groep"], df["totale_opbrengst"], color=["#d9534f", "#5bc0de"])

plt.ylabel("Totale opbrengst (€)")
plt.title("Opbrengst: klanten uit winkelstad vs. rest")

for bar, val in zip(bars, df["totale_opbrengst"]):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
             f"€{val:,.0f}", ha="center", va="bottom")
plt.tight_layout()
plt.show()
```
